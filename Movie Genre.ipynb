1.Read data Or Extract the data

```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
data=pd.read_csv("description.txt")
data
```

2.Create a Function to read data which is splited by ":::"

```python
def process_data(file_path):
    with open(file_path,'r',encoding='utf-8') as f:
        data=f.readlines()
    data=[line.strip().split(" ::: ") for line in data]

    return data

train_data=process_data("train_data.txt")
train_df=pd.DataFrame(train_data,columns=['ID','Title','Genre','Description'])

test_data=process_data("test_data.txt")
test_df=pd.DataFrame(test_data,columns=['ID','Title','Description'])

test_data_solution=process_data("test_data_solution.txt")
test_solution_df=pd.DataFrame(test_data_solution,columns=['ID','Title','Genre','Description'])

print("Train Data:")
train_df




    
```

```python
print("Test Data:")
test_df
```

```python
print("Test Dat solution:")
test_solution_df
```

3.Feature Extraction like TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer(max_features=10000)
x_train_tfidf=vectorizer.fit_transform(train_df["Description"])
x_test_tfidf=vectorizer.transform(test_df["Description"])
print(f"Training data shape: {x_train_tfidf.shape}")
print(f"Testing data shape: {x_test_tfidf.shape}")
```

4.Encoding the Target Labels

```python
from sklearn.preprocessing import LabelEncoder
label=LabelEncoder()
y_train=label.fit_transform(train_df['Genre'])
print(f"Genre in trainig data: {label.classes_}")
```

5.Model building using Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
LR_model=LogisticRegression(max_iter=200)
LR_model.fit(x_train_tfidf,y_train)
y_predict=LR_model.predict(x_test_tfidf)
predicted_genres=label.inverse_transform(y_predict)
test_df['Predicted_Genre']=predicted_genres
test_df[['Title','Predicted_Genre']]
```

```python
merged_df=pd.merge(test_solution_df[['ID','Genre']],test_df[['ID','Predicted_Genre']],on='ID')
merged_df
```

6.Model Evaluation using Logistic Regression

```python
from sklearn.metrics import accuracy_score,classification_report

accuracy=accuracy_score(merged_df['Genre'],merged_df['Predicted_Genre'])
print(f"Accuracy: {accuracy:.4f}")

print("\nClassification Report:")
print(classification_report(merged_df['Genre'],merged_df['Predicted_Genre']))
```
